# Vis for Search

## Keyword &rarr; documents

###  What's wrong with search (Hearst)

- Query responses don't tell you: 
    - How strong the match
    - How frequent each term
    - How term is distributed
    - Overlap between terms
    - Length of document
- Ranking is opaque
- Inability to compare results
- Imput limits term relationships

### TileBars (Hearst CHI'95)

- We've searched...
    ... can we say anything interesting about results?
- Objective: minimize time to decide which documents to look at

### TileBars: How they work



### TileBars: Advantages



- Simultaneous indication of: 
    - Relative document lengths
    - Frequency of term sets in document
    - Distribution of term sets with respect to the document and each other 

### TileBars: Negatives

- Mental model
- Linear vs non-linear materials
    - Images, flash tiles, etc.

### Takeaways:

- Visualizing search results is a common task
- Different sub-tasks 
    - What did we get back? Is it good?
    - What do I need to do to revise my search?
    - Can we use the vis to revice the search?
