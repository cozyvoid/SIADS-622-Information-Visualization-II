    - size indicating importance
    - the tags themselves

### Perception problems 

> Hearst '08
- Violates principles of perceptual design
    - Longer words grab more attention than shorter
        - Length of tag is conflated with its size
    - White space implies meaning when there is none intended
        - Ascenders and descenders can also effect focus
    - Eye moves around erratically, no flow or guides for visual focus
    - Proximity does not hold meaning
        - The paragraph-style payout makes it quite arbitary which termss are above, below, and otherwise near which other terms
    - Position within paragraph has saliency effects
    - Visual comparisons difficult (see Tufte)

### What are they good for?

- Signaling that there are tags
- Point of entry into complex site 
- Individual/group mirrors
- Fun

- See Hearst & Rosner, HICSS'08 for more 

1. Effective in (time constrained) category understanding tasks (vs. ungrouped layouts)
2. Grouping can be done by separating categories via white space, color or both
3. Organzied layouts with white space gaps preferred for analytic tasks (vs. tight packed, less organized looking)

Caveat: results for semantically distinct categories (others haven't been tested)
