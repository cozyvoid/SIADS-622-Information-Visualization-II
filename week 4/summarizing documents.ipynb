### Document metrics

- Literary analysis:
    - Word length
    - Syllables per word
    - Average sentence length
    - Percentage by parts of speech (nouns, verbs, etc.)
    - Frequencies of specific words
    - Hapax Legomena (words that appear once)

### Takeaways

- Document summarization is a common task for large collections
- Often requires abstracting documents or paragraphs into other mark types
- "Details on demand" can be used to recover actual text
