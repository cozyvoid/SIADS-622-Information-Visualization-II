# Network Data 

---

### The Language

- Node/Vertex
- Links 
	- Undirected edges (sometimes "edges")
		- Symmetric relationship (interacts-with, etc.)
	- Directed edges (sometimes "arcs" or "ties")
		- Asymmetric (parent-of, etc.)
- Graphs (set of nodes + set of links)
	- Types include:
		- directed/undirected/mixed
		- unweighted
		- multigraphs
		- forest
		- SCC
		- WCC
		- etc.

### Extra stuff

- In-degree
	- Number of inward pointing links to a node
- Out-degree
	- Number of outward pointing links
- Remember...
	- Nodes + Edges can have attributes 

### Network Vis (DOMAIN)

- Tasks
	- What do these two food Webs have in common (graph comparison)
	- Find duplicate nodes (e.g., ambiguous authors)
	- Have have friends been added over time (temporal evolution of the graph)

### Network Vis

- Tasks (ABSTRACTION)
	- Find adjacent nodes (count/get set/find max)
	- Find neighborhords (next door and further)
	- Find common neighbors
	- Find clusters, bridges, connected components
	- Find Nodes/edges with specific values/types
	- Follow a path 
	- etc.

- Techniques (TECHNIQUES)
	- Node-Link (many variants)
	- Hive Plots
	- Matrix

- Algorithms
	- Rendering the data
	- Aesthetic but also practical perceptual concerns
	- Scale - many layout algorithms are expensive and slow
	- Encoding additional information 

### Takeaways
	- Hierarchies are a sub-type of networks
	- Networks allow cycles
	- Networks and network components have many different names (depending on domain)
		- Recognition how these map is crucial 
	- As many tasks as there are domains 
	- Techniques are more limited
