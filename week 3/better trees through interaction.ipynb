### SpaceTree 











### Example: SpaceTree 

- Expressive: supports all tree-based comparisons, but can focus on ones of interest
- Effective:
	- Functional features: 
		- Single direction (mental model preservation)
		- Rescale branches to fit (data density)
		- Optimized camera movement (mental model preservation)
		- Preview icons (triangles)
		- Filter functions (ability to focus)

### Tradeoffs of SpaceTrees

- Tradeoff: 
	- Can focus on specific branch/entities (context lost)
	- Can see everything at same detail (focus lost) 
		- Effectiveness tradeoff
- Can we get back focus + context?
- Hyperbolic transformations
	- Fisheye/focus + context 

### Case 2

- Hyperbolic transformations
	- Fisheye/focus + context 
	
- Technique:
	- Lay out tree to hyperbolic plane
	- Map plane to disk 
	- Start at center - salience to focus area 
	- Animate to move in space - mental model preservation


> Lamping et al, 1995

- - Up to 10x as many nodes regular tree vis

### Key Features

- Magnification on center
- Layout only depends on 2-3 levels out
	- Cheaper rendering (don't have to draw far stuff)
- Smooth animation for change
