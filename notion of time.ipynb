## Temporal visualization (mostly time-series)

### Canonical (time-series) representation



"Of 4000 grapohic from 15 newspapers and magazines ('74-'78), 75% were time series."

- Tufte

### Data Mining vs. Data Vis.



- Algorithmic techniques
    - trends, patterns, etc.
    - good when objective is known
- Visualization
    - EDA
    - tasks
        - When was something at min/max?
        - Is there a pattern?
        - Two series similar?
        - Do any of the series match a pattern?
        - Provide fast access to the series.

### Other Tasks

- Does a data element exist at t?
- When does a data element exist?
- How long does it exist?
- How often does it exist?
- How fast is the data changing?
- Is what order do data elements appear?
- Do data elements exist together?

### Takeaways

- Temporal datasets are prevalent
- There are conventions for representing temporal data
- Often datsets are time + something else
- Need to  build a set of alternative representations
